```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

pd.options.display.max_columns = None

# Data Paths
BASE_PATH = Path("../../../data/aggregates")


def load_aggregates(base_path: Path):
    # Loading the three pillars of SkillCorner data
    physical = pd.read_csv(base_path / "aus1league_physicalaggregates_20242025.csv")
    obr = pd.read_csv(base_path / "aus1league_obraggregates_20242025.csv")
    passing = pd.read_csv(base_path / "aus1league_passingaggregates_20242025.csv")

    return physical, obr, passing


df_physical, df_obr, df_passing = load_aggregates(BASE_PATH)
print(
    f"Loaded {len(df_physical)} physical entries, {len(df_obr)} OBR entries, and {len(df_passing)} passing entries.",
)
```

```python
print(
    "Shapes\n"
    f"\tPhysical Aggregates: {df_physical.shape}\n"
    f"\tOff Ball Runs(OBR) Aggregates: {df_obr.shape}\n"
    f"\tPassing Aggregates: {df_passing.shape}",
)
```
